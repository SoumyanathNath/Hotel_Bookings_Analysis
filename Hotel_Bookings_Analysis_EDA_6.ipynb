In this dataset we get **112593** null value in 'company' column so we have replaced the null values with **0**

````python
#replace the null values 0.0 float

df['company'] = df['company'].fillna(0.0)
df['company'].head(23)
````

In the 'agent' column we have also get **16340** so we have replaced it with **0**

````python
#here also the value got replaced with 0.0

df['agent'] = df['agent'].fillna(0.0)
````

In country we can not set the country values as **0** becuase it contains country codes representing different countries. So instead of replacing it with zero we have replaced it with the mode value in country column. Mode is nothing but just the most repeating value.

````python
# Show all columns
pd.set_option('display.max_columns', None)

# Now this will show all columns
df.head()
````

````python
# To get the most frequently occurring country

df["country"].mode()[0]
````

````python
# FILL MISSING VALUES IN THE 'COUNTRY' COLUMN WITH THE MOST FREQUENT (MODE) VALUE
# Using [0] to select the first mode in case there are multiple

df["country"] = df["country"].fillna(df["country"].mode()[0])
````

````python
# CHILDREN COLUMN HAS THE COUNT OF CHILDREN THEN WE REPLACED MISSING VALUES WITH MEAN VALUES
# Fill missing values in the 'children' column with the column's mean value

df["children"] = df["children"].fillna(df["children"].mean())
````

````python
# Re-Count missing values

print(df.isnull().sum().sort_values(ascending=False))
````

````python
# IN THE DATA SET SOME ROWS HAVE ZERO GUESTS INCLUDING ADULTS, CHILDREN AND BABIES

df[(df.adults + df.babies + df.children)==0].shape
````

````python
''' SEVERAL ROWS IN THE DATASET CONTAINS VALUES THAT DOES NOT MAKE ANY SENSE LIKE HAVING NO ADULTS,
CHILDREN AND BABIES SO WE DIRECTLY DELETED IT BY USING DROP '''

df = df.drop(df[(df.adults + df.babies + df.children)==0].index)
df.shape
````

### 13. Night stay (Bar Chart)

````python
# Ensure df_not_canceled is a copy
df_not_canceled = df[df['is_canceled'] == 0].copy()

# Create new column 'total_nights'
df_not_canceled['total_nights'] = df_not_canceled['stays_in_weekend_nights'] + df_not_canceled['stays_in_week_nights']

# Plot
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(12, 6))
ax.set_xlabel('Total Nights Stayed')
ax.set_ylabel('Number of Bookings')
ax.set_title('Hotel-wise Night Stay Duration (Top 10 Most Common Durations)')

# Get top 10 most frequent 'total_nights' values
top_10_nights = df_not_canceled['total_nights'].value_counts().nlargest(10).index

# Plot countplot
sns.countplot(
    x='total_nights',
    hue='hotel',
    data=df_not_canceled[df_not_canceled['total_nights'].isin(top_10_nights)],
    order=top_10_nights,
    ax=ax
)

plt.show()
````

#####  What is/are the insight(s) found from the chart?

City Hotels dominate short stays (1–4 nights), while Resort Hotels attract relatively longer stays (around 7 nights).

The most common stay duration is 3 nights, indicating a preference for short trips.
